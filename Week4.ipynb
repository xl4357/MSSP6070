```python
import pandas as pd

df = pd.read_csv("/content/MSSP6070/MSSP6070/data/parsed.csv")

print("Dataset shape:", df.shape)

df.head()
```

```python
japan_mb = df[
    (df["parsed_place"] == "Japan") &
    (df["magType"] == "mb")
]

japan_95th_percentile = japan_mb["mag"].quantile(0.95)

print("95th percentile of earthquake magnitude in Japan (magType = mb):")
print(japan_95th_percentile)
```

```python
indonesia = df[df["parsed_place"] == "Indonesia"]

tsunami_percentage = indonesia["tsunami"].mean() * 100

print("Percentage of earthquakes in Indonesia coupled with tsunamis:")
print(tsunami_percentage)
```

```python
nevada = df[df["parsed_place"] == "Nevada"]

nevada_summary = nevada.describe()

print("Summary statistics for earthquakes in Nevada:")
nevada_summary
```

```python
ring_of_fire_locations = [
    "Bolivia",
    "Chile",
    "Ecuador",
    "Peru",
    "Costa Rica",
    "Guatemala",
    "Mexico",
    "Japan",
    "Philippines",
    "Indonesia",
    "New Zealand",
    "Canada",
    "Fiji",
    "Alaska",
    "Washington",
    "California",
    "Russia",
    "Taiwan",
    "Tonga",
    "Kermadec Islands"
]

df["ring_of_fire"] = (
    df["parsed_place"].isin(ring_of_fire_locations)
    | df["parsed_place"].str.contains("Antarctic", case=False, na=False)
)

df[["parsed_place", "ring_of_fire"]].head()
```
