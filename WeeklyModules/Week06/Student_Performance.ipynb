```python
import pandas as pd
```

```python
df = pd.read_csv("Student Performance Data.csv")
```

```python
print(df.head())
```

create lambda function

```python
classify_performance = lambda score: (
    "Excellent" if score >= 240
    else "Strong" if score >= 180
    else "Nice"
)
```

```python
df["total score"].apply(classify_performance)
```

Create new dataframe

```python
new_df = pd.DataFrame({
    "gender": df["gender"],
    "math score": df["math score"],
    "reading score": df["reading score"],
    "writing score": df["writing score"],
    "total score": df["total score"],
    "performance": df["total score"].apply(classify_performance)
})
```
