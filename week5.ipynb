```python
import pandas as pd
```

```python
aapl = pd.read_csv("MSSP6070/aapl.csv")
amzn = pd.read_csv("MSSP6070/amzn.csv")
fb = pd.read_csv("MSSP6070/fb.csv")
nflx = pd.read_csv("MSSP6070/nflx.csv")
goog = pd.read_csv("MSSP6070/goog.csv")
```

```python
aapl["ticker"] = "AAPL"
amzn["ticker"] = "AMZN"
fb["ticker"] = "FB"
nflx["ticker"] = "NFLX"
goog["ticker"] = "GOOG"
```

```python
aapl.head()
```

```python
stocks = pd.concat(
    [aapl, amzn, fb, nflx, goog],
    ignore_index=True
)
```

```python
stocks.to_csv("MSSP6070/data/faanf.csv", index=False)
```

```python
faanf = pd.read_csv("MSSP6070/data/faanf.csv")
```

```python
faanf["date"] = pd.to_datetime(faanf["date"])
faanf["volume"] = faanf["volume"].astype(int)
```

```python
faanf = faanf.sort_values(["date", "ticker"])
```

```python
faanf.nlargest(7, "volume")
```

```python
melted = faanf.melt(
    id_vars=["date", "ticker"],
    var_name="variable",
    value_name="value"
)

melted.head()
```
