## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
mean = sum(salaries)/len(salaries)
print(mean)
```

### median

```python
sorted_salaries = sorted(salaries)
median = (sorted_salaries[49]+sorted_salaries[50])/2
print(median)
```

### mode

```python
counts = {}
for salary in salaries:
  if salary in counts:
    counts[salary] += 1
  else:
    counts[salary] = 1
mode = max (counts, key=counts.get)
print(mode)
```

### sample variance
Remember to use Bessel's correction.

```python
total = 0
for salary in salaries:
    total = total + (salary - mean) ** 2
sample_variance = total / (len(salaries) - 1)
print(sample_variance)
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sample_standard_deviation = sample_variance ** 0.5
print(sample_standard_deviation)
```

## Exercise 6: Calculating more statistics
### range

```python
maximum = max(salaries)
minimum = min(salaries)
range = maximum - minimum
print(range)
```

### coefficient of variation

```python
coefficient_of_variation = sample_standard_deviation / mean
print(coefficient_of_variation)
```

### interquartile range

```python
sorted_salaries = sorted(salaries)
lower_half = sorted_salaries[:50]
upper_half = sorted_salaries[50:]
Q1 = (lower_half[24] + lower_half[25]) / 2
Q3 = (upper_half[24] + upper_half[25]) / 2
IQR = Q3 - Q1
print(IQR)
```

### quartile coefficent of dispersion

```python
sorted_salaries = sorted(salaries)
lower_half = sorted_salaries[:50]
upper_half = sorted_salaries[50:]
Q1 = (lower_half[24] + lower_half[25]) / 2
Q3 = (upper_half[24] + upper_half[25]) / 2
quartile_coefficient = (Q3 - Q1) / (Q3 + Q1)
print(quartile_coefficient)
```

## Exercise 7: Scaling data
### min-max scaling

```python
scaled_salaries = []
for salary in salaries:
    scaled = (salary - minimum) / (maximum - minimum)
    scaled_salaries.append(scaled)
print(scaled_salaries)
```

### standardizing

```python
standardized_salaries = []
for salary in salaries:
    x = (salary - mean) / sample_standard_deviation
    standardized_salaries.append(x)
print(standardized_salaries)
```
